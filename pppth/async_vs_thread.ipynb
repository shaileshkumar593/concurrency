```python
import asyncio
import time

async def worker(id: int):
    print(f"[Coroutine-{id}] Started")
    await asyncio.sleep(1)  # Simulate I/O-bound operation
    print(f"[Coroutine-{id}] Finished")

async def main():
    start = time.time()

    # Run 5 coroutines concurrently
    tasks = [asyncio.create_task(worker(i)) for i in range(1, 6)]
    await asyncio.gather(*tasks)

    end = time.time()
    print(f"✅ Total coroutine execution time: {end - start:.2f} seconds")

if __name__ == "__main__":
    asyncio.run(main())
```

```python
import threading
import time

def worker(id: int):
    print(f"[Thread-{id}] Started")
    time.sleep(1)  # Simulate I/O-bound operation
    print(f"[Thread-{id}] Finished")

def main():
    start = time.time()

    # Create and start 5 threads
    threads = [threading.Thread(target=worker, args=(i,)) for i in range(1, 6)]
    for t in threads:
        t.start()

    # Wait for all threads to complete
    for t in threads:
        t.join()

    end = time.time()
    print(f"✅ Total thread execution time: {end - start:.2f} seconds")

if __name__ == "__main__":
    main()
```

| Feature                      | Coroutines (Asyncio)               | Threads                     |
| ---------------------------- | ---------------------------------- | --------------------------- |
| **Concurrency Type**         | Cooperative (event loop)           | Preemptive (OS scheduler)   |
| **Context Switching**        | User-space, lightweight            | Kernel-level, heavier       |
| **Memory Usage**             | Very low (~few KB per task)        | Higher (~few MB per thread) |
| **Best For**                 | I/O-bound tasks (DB, network, API) | Mixed or CPU + I/O tasks    |
| **Blocking Calls**           | Must use `await` (non-blocking)    | Uses blocking I/O           |
| **Speed (5 tasks)**          | ~1s total                          | ~1s total                   |
| **Scalability (100k tasks)** | Handles easily                     | Will crash or freeze        |


 How it works:
asyncio.create_task() schedules all coroutines concurrently.

await asyncio.sleep() yields control to the event loop.

Despite 5 workers, total runtime ≈ 1 second, because all sleep concurrently.

How it works:
Each thread runs worker() concurrently.

The OS schedules threads preemptively.

Despite 5 threads sleeping for 1s, total runtime ≈ 1 second, same as coroutines.

Here’s what it actually does under the hood:

Creates a new event loop

It initializes the main event loop responsible for managing all async tasks.

Think of it as the "traffic controller" for your coroutines.

Runs the coroutine (main())

It executes the given coroutine (main) until completion.

Internally it schedules all awaited coroutines using that loop.

Closes the event loop cleanly

After the coroutine finishes, it automatically stops and closes the event loop — freeing up resources.

```python
import asyncio

async def say_hello():
    print("Hello ...")
    await asyncio.sleep(1)
    print("... World!")

async def main():
    await say_hello()

# Run the event loop
asyncio.run(main())
```

The main() function is declared as async, meaning it returns a coroutine.

asyncio.run(main()):

Creates an event loop

Executes main() until done

Cleans up and closes the loop
