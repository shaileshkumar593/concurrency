```python
l = [1,[2,[3,4]]]
result  = []

def flatten(l):
    for val in l:
        if isinstance(val, list):
            flatten(val)
        else:
            result.append(val)

    return result


print(flatten([1,[2,[3,4]]]))
```

```python
def flatten(lst):
    res = []
    for x in lst:
        if isinstance(x, list):
            res.extend(flatten(x))
        else:
            res.append(x)
    return res

print(flatten([1,[2,[3,4]]]))
```

```python
def unzip(lst):
    return list(zip(*lst))

print(unzip([(1,2),(3,4)]))
```

```python
from collections import Counter


def frequency (l):
    print(Counter(l))
    return dict(Counter(l))

ll = [1,1,2,3,3,3]

frequency(ll)
```

```python
# rotate matrix 90 degree

def rotate(mat):
    print(mat[::-1])
    return [list(x) for x in zip(*mat[::-1])]

print(rotate([[1,2],[3,4]]))
```

```python
lst = [[3,1],[2,2],[1]]


def process(lst):
    flat = [x for sub in lst for x in sub]
    return sorted(set(flat))


print(process(lst))
```

```python
def reverse_nested(lst):
    res = []
    for x in lst[::-1]:
        if isinstance(x, list):
            res.append(reverse_nested(x))
        else:
            res.append(x)
    return res

print(reverse_nested([1,[2,3],[4,[5]]]))
```
