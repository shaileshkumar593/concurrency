# multiprocessing 

```python
import time
import os
from PIL import Image, ImageFilter
import multiprocessing


filenames = [
    'images/1.jpg',
    'images/2.jpg',
    'images/3.jpg',
    'images/4.jpg',
    'images/5.jpg',
]


def create_thumbnail(filename, size=(50, 50), thumb_dir='thumbs'):
    try:
        # ensure output directory exists
        os.makedirs(thumb_dir, exist_ok=True)

        # open image safely
        with Image.open(filename) as img:
            # apply gaussian blur
            img = img.filter(ImageFilter.GaussianBlur())

            # create thumbnail
            img.thumbnail(size)

            # save thumbnail
            output_path = os.path.join(thumb_dir, os.path.basename(filename))
            img.save(output_path)

        print(f'{filename} was processed...')

    except Exception as e:
        print(f'Error processing {filename}: {e}')


def main():
    start = time.perf_counter()

    processes = []

    for filename in filenames:
        p = multiprocessing.Process(
            target=create_thumbnail,
            args=(filename,)
        )
        processes.append(p)
        p.start()

    for p in processes:
        p.join()

    finish = time.perf_counter()
    print(f'It took {finish - start:.2f} second(s) to finish')


if __name__ == '__main__':
    multiprocessing.freeze_support()  # ✅ Required on Windows
    main()
```

# processPoolExecutor 

```python
import time
import os
from PIL import Image, ImageFilter
from concurrent.futures import ProcessPoolExecutor
import multiprocessing


filenames = [
    'images/1.jpg',
    'images/2.jpg',
    'images/3.jpg',
    'images/4.jpg',
    'images/5.jpg',
]


def create_thumbnail(filename, size=(50, 50), thumb_dir='thumbs'):
    try:
        # ensure output directory exists
        os.makedirs(thumb_dir, exist_ok=True)

        # open image safely
        with Image.open(filename) as img:
            # apply gaussian blur
            img = img.filter(ImageFilter.GaussianBlur())

            # create thumbnail
            img.thumbnail(size)

            # save image
            output_path = os.path.join(thumb_dir, os.path.basename(filename))
            img.save(output_path)

        print(f'{filename} was processed...')

    except Exception as e:
        print(f'Error processing {filename}: {e}')


def main():
    start = time.perf_counter()

    with ProcessPoolExecutor() as executor:
        # Force evaluation to surface exceptions
        list(executor.map(create_thumbnail, filenames))

    finish = time.perf_counter()
    print(f'It took {finish - start:.2f} second(s) to finish')


if __name__ == '__main__':
    multiprocessing.freeze_support()  # ✅ REQUIRED on Windows
    main()
```
