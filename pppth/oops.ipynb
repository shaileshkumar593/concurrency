```python
# A class is a blueprint, and an object is an instance of that class.

# class_object_demo.py
"""
    __init__ is the constructor.

    self refers to the current instance.

    Objects are independent but share the same class structure.
"""

class Car:
    """A simple class representing a car."""

    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def start(self):
        print(f"{self.brand} {self.model} started!")

    def stop(self):
        print(f"{self.brand} {self.model} stopped.")


if __name__ == "__main__":
    car1 = Car("Tesla", "Model 3")
    car2 = Car("Toyota", "Corolla")

    car1.start()
    car2.start()
    car1.stop()
```

# Encapsulation hides internal details using private (__) and protected (_) attributes.
# __salary is private → cannot be accessed directly.
# Use getters/setters for controlled access.
# Promotes data hiding and safety.

```python
# encapsulation_demo.py

class Employee:
    def __init__(self, name, salary):
        self.name = name          # public
        self._bonus = 5000        # protected
        self.__salary = salary    # private

    def get_salary(self):
        return self.__salary

    def set_salary(self, new_salary):
        if new_salary > 0:
            self.__salary = new_salary
        else:
            print("Invalid salary amount")

if __name__ == "__main__":
    emp = Employee("Alice", 50000)
    print(emp.name)          # ✅ Accessible
    print(emp._bonus)        # ⚠️ Accessible but discouraged
    print(emp.get_salary())  # ✅ Access via getter

    emp.set_salary(60000)
    print(emp.get_salary())
```

# Inheritance allows one class to reuse attributes/methods of another.

```python
# inheritance_demo.py

class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return "Makes a sound"


class Dog(Animal):
    def speak(self):
        return f"{self.name} says Woof!"


class Cat(Animal):
    def speak(self):
        return f"{self.name} says Meow!"


if __name__ == "__main__":
    dog = Dog("Rex")
    cat = Cat("Misty")

    print(dog.speak())
    print(cat.speak())
```

# Polymorphism = many forms → same interface, different behavior.
# All objects share the same fly() interface.
# The right method is chosen dynamically at runtime.

```python
# polymorphism_demo.py

class Bird:
    def fly(self):
        print("Some birds can fly!")

class Sparrow(Bird):
    def fly(self):
        print("Sparrow flies high!")

class Penguin(Bird):
    def fly(self):
        print("Penguins cannot fly!")

def show_flight(bird: Bird):
    bird.fly()

if __name__ == "__main__":
    birds = [Sparrow(), Penguin(), Bird()]
    for b in birds:
        show_flight(b)
```

# Abstraction hides unnecessary details using abstract base classes.
# Vehicle is abstract → cannot be instantiated.
# Subclasses must implement abstract methods.
# Encourages interface-driven design.

```python
# abstraction_demo.py

from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass

class Bike(Vehicle):
    def start(self):
        print("Bike started...")

    def stop(self):
        print("Bike stopped...")

if __name__ == "__main__":
    bike = Bike()
    bike.start()
    bike.stop()
```

# Class Methods & Static Methods

```python
# class_static_demo.py

class MathUtils:
    version = "1.0"

    def __init__(self):
        pass

    @staticmethod
    def add(a, b):
        return a + b

    @classmethod
    def get_version(cls):
        return f"MathUtils v{cls.version}"

if __name__ == "__main__":
    print(MathUtils.add(5, 10))       # Static method
    print(MathUtils.get_version())    # Class method
```

# Operator Overloading
# Python lets you redefine built-in operators (e.g. +, -, ==).

```python
# operator_overloading_demo.py

class Money:
    def __init__(self, amount):
        self.amount = amount

    def __add__(self, other):
        return Money(self.amount + other.amount)

    def __str__(self):
        return f"₹{self.amount}"

if __name__ == "__main__":
    m1 = Money(100)
    m2 = Money(200)
    print(m1 + m2)   # Calls __add__
```

# Composition (HAS-A Relationship)
# Composition means a class contains another class as an attribute.

```python
# composition_demo.py

class Engine:
    def start(self):
        print("Engine started!")

class Car:
    def __init__(self, brand):
        self.brand = brand
        self.engine = Engine()  # Composition

    def drive(self):
        self.engine.start()
        print(f"{self.brand} is driving...")

if __name__ == "__main__":
    car = Car("BMW")
    car.drive()
```
