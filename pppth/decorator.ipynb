```python
from functools import wraps

class RateLimitException(Exception):
    pass


def rate_limit(max_calls):
    def decorator(func):
        count = 0

        @wraps(func)
        def wrapper(*args, **kwargs):
            nonlocal count

            if count >= max_calls:
                raise RateLimitException("Limit exceeded")

            count += 1
            return func(*args, **kwargs)

        return wrapper
    return decorator


@rate_limit(5)
def print_hello():
    print("hello")


for i in range(7):
    try:
        print(f"Call {i+1}: ", end="")
        print_hello()
    except RateLimitException as e:
        print(e)
```

```python
#Basic Decorator

from functools import wraps

def my_decorator(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        print("Before function")
        result = func(*args, **kwargs)
        print("After function")
        return result
    return wrapper


@my_decorator
def greet():
    print("Hello")


greet()
```

```python
# Decorator With Argument

from functools import wraps

def repeat(n):
    def decorator(func):
        @wraps(func)
        def wrapper(*args, **kwargs):
            for _ in range(n):
                func(*args, **kwargs)
        return wrapper
    return decorator


@repeat(3)
def say_hello():
    print("Hello")


say_hello()
```

```python
# Function With Arguments (Decorator Handles Params)

from functools import wraps

def log_args(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        print(f"Args: {args}, Kwargs: {kwargs}")
        return func(*args, **kwargs)
    return wrapper


@log_args
def add(a, b):
    return a + b


print(add(2, 3))
```

```python
# Class-Based Decorator
class CountCalls:
    def __init__(self, func):
        self.func = func
        self.count = 0

    def __call__(self, *args, **kwargs):
        self.count += 1
        print(f"Called {self.count} times")
        return self.func(*args, **kwargs)


@CountCalls
def greet():
    print("Hello")


greet()
greet()
```
