An iterable is any Python object you can loop over (e.g., with a for loop).
It implements the __iter__() method, which returns an iterator.

✅ Examples of iterables:
list_data = [1, 2, 3]
tuple_data = (4, 5, 6)
string_data = "abc"
set_data = {7, 8, 9}
dict_data = {"a": 1, "b": 2}

An iterator is the object that actually performs the iteration — it keeps track of the current position and knows how to fetch the next item.

It implements two methods:

__iter__() → returns itself

__next__() → returns the next item (and raises StopIteration when done)

You can get an iterator from an iterable using iter().

```python
nums = [10, 20, 30]   # iterable
it = iter(nums)       # create iterator

print(next(it))  # 10
print(next(it))  # 20
print(next(it))  # 30
print(next(it))  # Raises StopIteration
```

Python overallocates capacity in chunks — typically grows by ~1.125× factor.
Estimating “capacity”
Although Python doesn’t give cap() directly, you can approximate it using:
approx_capacity = (sys.getsizeof(nums) - 56) // 8


Explanation:

Each Python list object has a header (~56 bytes on 64-bit systems).

Each pointer (to element) costs ~8 bytes.

So the difference gives approximate number of pointer slots (capacity).

```python
import sys

nums = []
print(f"{'len':>3} {'bytes':>8} {'approx_cap':>12}")
for i in range(20):
    nums.append(i)
    cap = (sys.getsizeof(nums) - 56) // 8
    print(f"{len(nums):3d} {sys.getsizeof(nums):8d} {cap:12d}")
```

# list reversal 

```python
my_list = [1, 2, 3, 4, 5]
reversed_list = my_list[::-1]
print(reversed_list) # Output: [5, 4, 3, 2, 1]
```

```python
my_list = [10, 20, 30, 40, 50, 60]
sub_list_reversed = my_list[4:1:-1] # From index 4 down to (but not including) index 1
print(sub_list_reversed) # Output: [50, 40, 30]
```

```python
my_list = [1, 2, 3, 4, 5]
empty_slice = my_list[1:4:-1]
print(empty_slice) # Output: []
```

```python
# Using the reverse() method (in-place modification):
my_list = [1, 2, 3, 4, 5]
my_list.reverse()
print(my_list)
# Output: [5, 4, 3, 2, 1]
```

```python
original_list = [1, 2, 3, 4, 5]
reversed_list = original_list[::-1]
print(original_list)
# Output: [1, 2, 3, 4, 5]
print(reversed_list)
# Output: [5, 4, 3, 2, 1]
```

```python
my_list = [1, 2, 3, 4, 5]
reversed_iterator = reversed(my_list)
print(reversed_iterator)

# To get a reversed list:
reversed_list_from_iterator = list(reversed_iterator)
print(my_list)
# Output: [1, 2, 3, 4, 5]
print(reversed_list_from_iterator)
# Output: [5, 4, 3, 2, 1]
```

sort() Syntax
numbers.sort(reverse, key)
The sort() method can take two optional keyword arguments:

reverse - By default False. If True is passed, the list is sorted in descending order.
key - Comparion is based on this function.

```python
text = ["abc", "wxyz", "gh", "a"]

# stort strings based on their length
text.sort(key = len)
print(text)
```

```python
cities = ["Tokyo", "London", "Washington D.C"]

# sort in dictionary order
cities.sort()
print(f"Dictionary order: {cities}")

# sort in reverse dictionary order
cities.sort(reverse = True)
print(f"Reverse dictionary order: {cities}")
```

In Python, a list is an implementation of a dynamic array. Unlike static arrays in other languages (like C) which have a fixed size defined at creation, Python lists can grow and shrink as elements are added or removed during runtime. 

How Python Lists Act as Dynamic Arrays
The "dynamic" behavior is managed automatically by the Python interpreter through a process of over-allocation and resizing: 
Initial Allocation: When a list is created, Python allocates a contiguous block of memory that is slightly larger than the current number of elements required to store pointers to the actual data objects. This extra space acts as "spare room" for future additions.
Appending Elements (Common Case): Most append() operations are fast (O(1) time complexity) because the new element can be added directly into the pre-allocated spare memory slot.
Automatic Resizing (When Full): When all the allocated space is consumed and an additional element is appended, Python performs a resizing operation:
It allocates a new, larger block of memory (typically 1.125 to 2 times the current size, depending on the implementation).
It copies all the elements (specifically, their pointers) from the old memory block to the new, larger block.
It then deallocates the old memory block.
Finally, the new element is added to the newly available space.
Amortized Efficiency: Although an individual resize operation is expensive (O(n) time complexity, where n is the number of elements copied), this process happens infrequently because of the over-allocation strategy. When averaged over many append operations, the amortized time complexity of list.append() is O(1)

# * and ** uses 

```python
def add(*args):
    return sum(args)

print(add(1, 2, 3, 4))
```

```python
nums = [1, 2, 3]

print(*nums)
```

```python
# passing parameter
def multiply(a, b, c):
    return a * b * c

nums = [2, 3, 4]

print(multiply(*nums))
```

```python
a = [1, 2]
b = [3, 4]

c = [*a, *b]
print(c)
```

```python
def func(a, *, b):
    print(a, b)

func(1, b=2)
```

# **kwargs → Variable Keyword Arguments

```python
def user(**kwargs):
    print(kwargs)

user(name="Ram", age=25)
```

```python
d = {"a": 1, "b": 2}

def func(a, b):
    print(a, b)

func(**d)
```
