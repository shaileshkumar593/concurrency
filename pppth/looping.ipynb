| Term         | Description                                                           | Example                               |
| ------------ | --------------------------------------------------------------------- | ------------------------------------- |
| **Iterable** | Any object that can return an iterator (supports `__iter__()`).       | list, tuple, dict, set, string, range |
| **Iterator** | Object that produces data one item at a time (supports `__next__()`). | Returned by calling `iter(iterable)`  |
| **range()**  | Built-in iterable that generates numbers lazily (iterator-like).      | `range(5)` yields 0 → 4               |


Iteration on List (Iterable → Iterator)
✅ Example 1: Manual Iteration using iter() and next()

```python
nums = [10, 20, 30, 40]

# Convert list (iterable) → iterator
it = iter(nums)

print(next(it))  # 10
print(next(it))  # 20
print(next(it))  # 30
print(next(it))  # 40
# print(next(it))  # StopIteration (if uncommented)
```

```python
#Example 2: Using for Loop (Python automatically uses iterator internally)
nums = [10, 20, 30, 40]

for n in nums:  # internally calls iter(nums) and repeatedly next()
    print(n)
```

```python
it = iter(nums)
while True:
    try:
        n = next(it)
        print(n)
    except StopIteration:
        break
```

```python
# Example 3: Using range() on list indices
nums = [10, 20, 30, 40]

for i in range(len(nums)):
    print(f"Index {i} → Value {nums[i]}")
```

Iteration on Dictionary
✅ Example 1: Loop over keys (default)

```python
data = {"name": "Alice", "age": 25, "city": "Delhi"}

for key in data:
    print(key, ":", data[key])
```

```python
# Loop using .items() (key-value pair)

for key, value in data.items():
    print(f"{key} → {value}")
```

```python
# Using Iterator explicitly

data = {"name": "Alice", "age": 25, "city": "Delhi"}

it = iter(data.items())

print(next(it))  # ('name', 'Alice')
print(next(it))  # ('age', 25)
print(next(it))  # ('city', 'Delhi')
```

```python
# Using range() for dictionary keys (with list conversion)

data = {"a": 100, "b": 200, "c": 300}
keys = list(data.keys())

for i in range(len(keys)):
    key = keys[i]
    print(f"{i}. {key} = {data[key]}")
```
