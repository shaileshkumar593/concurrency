```python
# math_decorator_simple.py

def math_decorator(operation_name):
    """
    A simple decorator factory that accepts an operation name
    and decorates arithmetic functions.
    """
    def decorator(func):
        def wrapper(a, b):
            print(f"🔹 Performing {operation_name.upper()} Operation")
            print(f"➡️ Inputs: a = {a}, b = {b}")

            # Execute the actual function
            result = func(a, b)

            print(f"✅ Result: {result}\n")
            return result
        return wrapper
    return decorator


@math_decorator("addition")
def add(a, b):
    return a + b


@math_decorator("subtraction")
def sub(a, b):
    return a - b


@math_decorator("multiplication")
def mul(a, b):
    return a * b


# ---------- Main Execution ----------
if __name__ == "__main__":
    add(10, 5)
    sub(10, 5)
    mul(10, 5)
```

# Let’s break this step-by-step (with execution flow visualization) so you understand exactly what happens when you use:

@math_decorator("addition")
def add(a, b):
    return a + b
🧩 Step 1 — Function Definition Time
When Python sees the decorator syntax @math_decorator("addition"),
it doesn’t immediately run add() — instead it executes the decorator definition chain.

This line:
@math_decorator("addition")
is equivalent to this:

add = math_decorator("addition")(add)
Let’s unwrap that 👇

⚙️ Step 2 — The Call Chain
1️⃣ math_decorator("addition") is called first.
This executes the outermost function and returns a new function decorator.

math_decorator("addition") → returns decorator
So now, Python has:

add = decorator(add)
2️⃣ Next, the decorator(add) function is called.
This executes the middle function, which takes add as its argument and returns another function — the wrapper.

decorator(add) → returns wrapper
So now:

add = wrapper
✅ The original add function is replaced by the wrapper function.

3️⃣ The final add() you call at runtime is actually wrapper(a, b).
When you do:

add(10, 5)
It’s equivalent to:

wrapper(10, 5)
And inside the wrapper:

def wrapper(a, b):
    print("Before calling original function")
    result = func(a, b)  # func = original add
    print("After calling original function")
    return result
So the call flow becomes:

main() → add(10, 5)
          ↓
       wrapper(10, 5)
          ↓
       func(a, b)  → original add(10, 5)

```python
def math_decorator(operation_name):
    def decorator(func):
        def wrapper(a, b):
            if not (isinstance(a, (int, float)) and isinstance(b, (int, float))):
                print("❌ Error: Both inputs must be numbers.\n")
                return None

            print(f"🔹 Performing {operation_name.upper()} Operation")
            print(f"➡️ Inputs: a = {a}, b = {b}")

            result = func(a, b)
            print(f"✅ Result: {result}\n")
            return result
        return wrapper
    return decorator


@math_decorator("addition")
def add(a, b): return a + b

@math_decorator("subtraction")
def sub(a, b): return a - b

@math_decorator("multiplication")
def mul(a, b): return a * b


if __name__ == "__main__":
    add(10, 2)
    sub(15, 3)
    mul(4, 5)
    add("x", 3)   # Triggers validation
```

```python
def my_decorator(func):
    def wrapper():
        print("Before function call")
        func()
        print("After function call")
    return wrapper

@my_decorator
def say_hello():
    print("Hello!")

say_hello()
```

# 3. How the Call Actually Happens
# Let’s look at this step-by-step:

# @my_decorator
# def say_hello():
#    print("Hello!")
# is equivalent to:

# def say_hello():
#    print("Hello!")

# say_hello = my_decorator(say_hello)
# So now, say_hello actually refers to wrapper returned by the decorator.

# So when you call say_hello(), Python is really doing:

#            wrapper()
# which internally calls the original func() inside.

```python
def operation_logger(func):
    def wrapper(a, b):
        print(f"Calling function: {func.__name__}")
        print(f"Inputs → a: {a}, b: {b}")
        result = func(a, b)
        print(f"Result → {result}")
        print("-" * 30)
        return result
    return wrapper

@operation_logger
def add(a, b):
    return a + b

@operation_logger
def sub(a, b):
    return a - b

@operation_logger
def mul(a, b):
    return a * b


# Calls
add(10, 5)
sub(10, 5)
mul(10, 5)
```
