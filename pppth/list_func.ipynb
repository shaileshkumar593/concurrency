```python
# mixed list
# The copy() method returns a new list. It doesn't modify the original list.
# If you modify the new_list in the above example, my_list will not be modified.


my_list = ['cat', 0, 6.7]

# copying a list
new_list = my_list.copy()

print('Copied List:', new_list)

new_list[0]=99
print(my_list)
print(new_list)
```

```python
def my_sum(*args):
    result = 0
    for num in args:
        result += num
    return result

print(my_sum(1, 2, 3, 4, 5)) # Output: 15
```

```python
def print_details(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_details(name="John", age=30, country="USA")
# Output:
# name: John
# age: 30
# country: USA
```

```python
def my_function(arg1, arg2, arg3):
    print(f"arg1: {arg1}, arg2: {arg2}, arg3: {arg3}")

args_tuple = ("two", 3, 5)
my_function(*args_tuple) # Unpacks the tuple into positional arguments
# Output: arg1: two, arg2: 3, arg3: 5

kwargs_dict = {"arg3": 3, "arg2": "two", "arg1": 5}
my_function(**kwargs_dict) # Unpacks the dictionary into keyword arguments
# Output: arg1: 5, arg2: two, arg3: 3
```
