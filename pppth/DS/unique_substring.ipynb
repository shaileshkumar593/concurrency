```python
def UniqueLongestSubstrs(a):
    s = set()
    l = 0
    max_so_far = 0
    start = 0   # start index of longest substring

    for r in range(len(a)):
        while a[r] in s:
            s.remove(a[l])
            l += 1

        s.add(a[r])

        if r - l + 1 > max_so_far:
            max_so_far = r - l + 1
            start = l

    print("Longest Unique Substring:", a[start:start + max_so_far])
    return max_so_far


print("Length:", UniqueLongestSubstrs("abcabcbb"))
```
