```python
from contextlib import contextmanager

@contextmanager
def open_file(filename, mode):
    file = open(filename, mode)
    try:
        yield file # Pauses the function and returns the resource
    finally:
        file.close() # Cleanup logic runs after the with block finishes

with open_file("example.txt", "r") as file:
    content = file.read()
    print(content)
```
