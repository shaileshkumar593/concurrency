A Python set is an unordered collection of unique elements. It is a mutable data type, meaning elements can be added or removed after creation. Sets are particularly useful for tasks involving membership testing, removing duplicates from sequences, and performing mathematical set operations. 

Key Characteristics of Python Sets:
# Unordered: Elements in a set do not have a defined order and cannot be accessed by index.
# Unique Elements: Sets automatically handle duplicate values, ensuring each element within a set is distinct.
# Mutable: Elements can be added or removed from a set after it has been created.
# Hashable Elements: Elements stored in a set must be hashable, meaning they must be immutable (like numbers, strings, or tuples). Lists and # dictionaries cannot be directly stored as elements within a set because they are mutable.

```python
# Using curly braces
my_set = {1, 2, 3, 4, 5}

# Using the set() constructor with an iterable (e.g., a list)
another_set = set([1, 2, 2, 3, 4, 5]) # Duplicates are automatically removed
```

```python
# create a set of integer type
student_id = {112, 114, 116, 118, 115}
print('Student ID:', student_id)

# create a set of string type
vowel_letters = {'a', 'e', 'i', 'o', 'u'}
print('Vowel Letters:', vowel_letters)

# create a set of mixed data types
mixed_set = {'Hello', 101, -2, 'Bye'}
print('Set of mixed data types:', mixed_set)
```

```python
# create an empty set
empty_set = set()

# create an empty dictionary
empty_dictionary = { }

# check data type of empty_set
print('Data type of empty_set:', type(empty_set))

# check data type of dictionary_set
print('Data type of empty_dictionary:', type(empty_dictionary))
```

# duplicate element in set is removed on running 

```python
numbers = {2, 4, 6, 6, 2, 8}
print(numbers)   # {8, 2, 4, 6}
```

# set function   

```python
numbers = {21, 34, 54, 12}

print('Initial Set:',numbers)

# using add() method
numbers.add(32)

print('Updated Set:', numbers) 
```

```python
#  update() method is used to update the set with items other collection types (lists, tuples, sets, etc)
companies = {'Lacoste', 'Ralph Lauren'}
tech_companies = ['apple', 'google', 'apple']

# using update() method
companies.update(tech_companies)

print(companies)

# Output: {'google', 'apple', 'Lacoste', 'Ralph Lauren'}
```

```python
#  discard() method to remove the specified element from a set
languages = {'Swift', 'Java', 'Python'}

print('Initial Set:',languages)

# remove 'Java' from a set
removedValue = languages.discard('Java')

print('Set after remove():', languages)
```

A frozenset in Python is an immutable version of a set. Like a regular set, it is an unordered collection of unique elements. The key distinction is its immutability: once a frozenset is created, its elements cannot be added, removed, or modified. 


Key Characteristics:
# Immutable: Elements cannot be changed after creation. This means operations like add(), remove(), or discard() are not available for frozenset objects.
# Hashable: Because frozenset objects are immutable, they are also hashable. This allows them to be used as elements in other sets or as keys in dictionaries, which is not possible with mutable set objects.
# Unordered: The order of elements in a frozenset is not guaranteed and may vary.
# Unique Elements: Like regular sets, frozenset objects only store unique elements; duplicates are automatically removed during creation.

```python
# Creating a frozenset from a list
my_list = [1, 2, 3, 2, 1]
frozen_set = frozenset(my_list)
print(frozen_set)

# Creating an empty frozenset
empty_frozen_set = frozenset()
print(empty_frozen_set)
```

```python
# Create a frozenset from a list
my_list = [1, 2, 3, 4, 3]
frozen_set = frozenset(my_list)
print(frozen_set)
# Output: frozenset({1, 2, 3, 4})

# Create a frozenset from a tuple
my_tuple = ('apple', 'banana', 'cherry')
frozen_set2 = frozenset(my_tuple)
print(frozen_set2)
# Output: frozenset({'apple', 'banana', 'cherry'})

# Attempting to modify a frozenset raises an error
try:
    frozen_set.add(5)
except AttributeError as e:
    print(f"Error: {e}")
# Output: Error: 'frozenset' object has no attribute 'add'
```

Common Use Cases:
# Dictionary Keys: When you need a collection of unique items as a key in a dictionary.
# Elements in another Set: When you need a collection of unique items to be an element within another set.
# Ensuring Data Integrity: When you want to guarantee that a collection of unique items remains unchanged throughout your program.


Function	Description
| **Function / Method**           |      **Description**                                                                                     **
| ------------------------------- | ------------------------------------------------------------------------------------------------------------------------------- |
| `all()`                         | Returns `True` if all elements of the set are true (or if the set is empty).                                                    |
| `any()`                         | Returns `True` if any element of the set is true. If the set is empty, returns `False`.                                         |
| `enumerate()`                   | Returns an enumerate object containing index–value pairs of the set items. *(Note: sets are unordered, so order is arbitrary.)* |
| `len()`                         | Returns the number of items in the set.                                                                                         |
| `max()`                         | Returns the largest item in the set.                                                                                            |
| `min()`                         | Returns the smallest item in the set.                                                                                           |
| `sorted()`                      | Returns a new sorted list from elements in the set (does not modify the set).                                                   |
| `sum()`                         | Returns the sum of all numeric elements in the set.                                                                             |
| `add()`                         | Adds a single element to the set.                                                                                               |
| `clear()`                       | Removes all elements from the set.                                                                                              |
| `copy()`                        | Returns a shallow copy of the set.                                                                                              |
| `difference()`                  | Returns a new set containing elements present in this set but not in others.                                                    |
| `difference_update()`           | Removes all elements of another set from this set (updates in place).                                                           |
| `discard()`                     | Removes an element from the set if it exists; does nothing if not found.                                                        |
| `intersection()`                | Returns a new set containing elements common to all sets.                                                                       |
| `intersection_update()`         | Updates this set with elements common to all sets.                                                                              |
| `isdisjoint()`                  | Returns `True` if two sets have no elements in common.                                                                          |
| `issubset()`                    | Returns `True` if this set is contained within another set.                                                                     |
| `issuperset()`                  | Returns `True` if this set contains all elements of another set.                                                                |
| `pop()`                         | Removes and returns an arbitrary element from the set; raises `KeyError` if empty.                                              |
| `remove()`                      | Removes a specified element from the set; raises `KeyError` if not found.                                                       |
| `symmetric_difference()`        | Returns a new set containing elements in either set but not in both.                                                            |
| `symmetric_difference_update()` | Updates this set with elements that are in either set but not both.                                                             |
| `union()`                       | Returns a new set containing all unique elements from all provided sets.                                                        |
| `update()`                      | Updates this set with the union of itself and others.                                                                           |
